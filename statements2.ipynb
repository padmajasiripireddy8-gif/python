```python
#sum of digits in a number 
number = int(input("Enter number:"))

total = 0

while number > 0:
    digit = number % 10
    total = total + digit
    number = number // 10
    
print("Sum of digits:", total)
```

```python
#reverse a number 
number = int(input("Enter number: "))

reverse = 0

while number > 0:
    digit = number % 10
    number = number // 10
    reverse = reverse * 10 + digit 
    
print("Reverse:", reverse)
```

```python
#check if a number is a palindrome
number = int(input("Enter number: "))

original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 30
    
if original == reverse:
    print("palindrome")
    
else:
    print("Not palindrome")
```

```python
#check if number is prime
number = int(input("Enter number:"))

count = 0

for i in range(1, number + 1):
    if number % i == 0:
        count = count + 1
        
if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```
