```python
#tuples in  python
#tuple is a collection which is ordered and unchangeable. In python tuples are written with round brackets.
student = ("Jayasree",98,"python")

print(student[0])
```

```python
#tuples are immutable , meaning they cannot be changed after they are created.they are defined
numbers = (10, 20, 20, 30, 20)

print(numbers.count(20))
```

```python
numbers = (10,20,30,40)
print(numbers.index(30))
```

```python
numbers = (10, 20, 30, 40,)

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

```python
#sets in python 
#Set is a collection of unique values that is unordered and unindexed. Sets are defined by values separated by commas inside curly braces {}.{}

numbers = {10, 20, 30, 20, 10}

print(numbers)
```

```python
#add values to a set
subjects = {"Python", "Java"}

subjects.add("SQL")
print(subjects)

```

```python
#remove values from a set 
subjects.remove("Java")

print(subjects)
```

```python
#set do not allow dupllicate values
numbers = {1, 2, 2, 3, 3, 4}

print(numbers)
```

```python
#dictionaries in python 
#Dictionary is a collection of unique key value pairs that is unordered and changable
student = {
    "name": "Jayasree",
    "age": 18,
    "course": "python"}

print(student)
    

     
```

```python
#access elements in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"] = "vijayawada"

print(student)
```

```python
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
student.update({"age": 20})

print(student)

student.pop("age")

print(student)
```

```python
#popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead).
student = {
    "name": "Jayasree",
    "age": 18,
    "course": "python"
}

student.popitem()

print(student)
```

```python
student = {
    "name": "Jayasree",
}

student.setdefault("age", 18)

print(student)
```
